```python
from pathlib import Path
from urllib.request import urlopen
import ssl
import certifi
import pandas as pd

raw_folder = Path("../data/raw")
raw_folder.mkdir(parents=True, exist_ok=True)

url = "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2017/DataFiles/DPQ_J.xpt"
file_path = raw_folder / "DPQ_J.XPT"

# Use trusted certificates
ssl_context = ssl.create_default_context(cafile=certifi.where())

with urlopen(url, context=ssl_context) as response:
    file_path.write_bytes(response.read())

depression_df = pd.read_sas(file_path, format="xport")

print("Download successful!")
print("Dataset shape:", depression_df.shape)

depression_df.head()
```

```python
import pyreadstat

depression_df, metadata = pyreadstat.read_xport(str(file_path))

print("Dataset shape:", depression_df.shape)
depression_df.head()
```

```python
from functools import reduce

# Store all source datasets in one list
source_datasets = [
    depression_df,
    demographics_df,
    sleep_df,
    body_df,
    activity_df,
    smoking_df,
    alcohol_df
]

# Verify that each participant ID appears only once in each dataset
for dataset in source_datasets:
    assert dataset["SEQN"].is_unique, "Duplicate SEQN values detected."

# Merge all datasets with the depression sample as the starting population
# A left join preserves all 5,533 depression-questionnaire participants
merged_df = reduce(
    lambda left, right: left.merge(
        right,
        on="SEQN",
        how="left",
        validate="one_to_one"
    ),
    source_datasets
)

# Confirm the merged dataset's size and participant uniqueness
print("Merged dataset shape:", merged_df.shape)
print("Unique participants:", merged_df["SEQN"].nunique())
print("Duplicate participants:", merged_df["SEQN"].duplicated().sum())

# Display the first five merged records
merged_df.head()
```

```python
# Depression questionnaire items used only to create the target variable
depression_columns = [
    "DPQ010", "DPQ020", "DPQ030",
    "DPQ040", "DPQ050", "DPQ060",
    "DPQ070", "DPQ080", "DPQ090"
]

# Demographic and socioeconomic variables
demographic_columns = [
    "RIAGENDR",   # Gender
    "RIDAGEYR",   # Age in years
    "RIDRETH3",   # Race and ethnicity
    "DMDEDUC2",   # Education level
    "DMDMARTL",   # Marital status
    "INDFMPIR",   # Family income-to-poverty ratio
    "WTMEC2YR",   # NHANES examination weight
    "SDMVPSU",    # Survey sampling unit
    "SDMVSTRA"    # Survey sampling stratum
]

# Sleep-related variables
sleep_columns = [
    "SLD012",     # Weekday sleep duration
    "SLD013",     # Weekend sleep duration
    "SLQ030",     # Trouble sleeping
    "SLQ040",     # Diagnosed sleep disorder
    "SLQ050",     # Snoring frequency
    "SLQ120"      # Daytime sleepiness
]

# Body-measurement variables
body_columns = [
    "BMXBMI",     # Body mass index
    "BMXWAIST"    # Waist circumference
]

# Physical-activity and sedentary-behavior variables
activity_columns = [
    "PAQ605",     # Vigorous work activity
    "PAQ620",     # Moderate work activity
    "PAQ635",     # Walking or bicycling
    "PAQ650",     # Vigorous recreational activity
    "PAQ665",     # Moderate recreational activity
    "PAD680"      # Minutes of sedentary activity
]

# Smoking-related variables
smoking_columns = [
    "SMQ020",     # Smoked at least 100 cigarettes
    "SMQ040",     # Current smoking status
    "SMD650"      # Average cigarettes per day
]

# Alcohol-use variables
alcohol_columns = [
    "ALQ111",
    "ALQ121",
    "ALQ130",
    "ALQ142",
    "ALQ151",
    "ALQ170"
]

# Combine the identifier and all selected variable groups
selected_columns = (
    ["SEQN"]
    + depression_columns
    + demographic_columns
    + sleep_columns
    + body_columns
    + activity_columns
    + smoking_columns
    + alcohol_columns
)

# Stop with a clear error if any requested column is unavailable
missing_columns = [
    column for column in selected_columns
    if column not in merged_df.columns
]

if missing_columns:
    raise KeyError(f"Missing columns: {missing_columns}")

# Create the smaller analysis dataset
analysis_df = merged_df[selected_columns].copy()

# Confirm the new dataset dimensions
print("Selected dataset shape:", analysis_df.shape)

# Display the first five records
analysis_df.head()
```

```python
# Create a folder for merged but not-yet-cleaned data
interim_folder = Path("../data/interim")
interim_folder.mkdir(parents=True, exist_ok=True)

# Define the output file location
interim_file = interim_folder / "nhanes_2017_2018_merged.csv"

# Save the selected dataset without the pandas index
analysis_df.to_csv(interim_file, index=False)

# Confirm that the file was created successfully
print("Dataset saved successfully!")
print("Saved location:", interim_file)
print("Saved shape:", analysis_df.shape)
```
